## 1. Archive Exploration

This [Water Conflict and Crisis Events in the Colorado River Basin](https://www.sciencebase.gov/catalog/item/63acac09d34e92aad3ca1480) dataset from the U.S. Geological Survey (USGS) contains data collected from 1) water control board and official organization public meetings regarding the Colorado River Basin and 2) relevant article and newspaper reviews. Data collection spans the years 1970-2005, with the goal of better elucidating water insecurity and conflict as they relate to social vulnerability and improved resource management in the Basin area. 

D. V. Holloman, M. K. Hines, and D. K. Zoanni, “Coded Water Conflict and Crisis Events in the Colorado River Basin, Derived from LexisNexis search 2005-2021.” U.S. Geological Survey, 2023. doi: 10.5066/P9X6WR7J. Available: [https://www.sciencebase.gov/catalog/item/63acac09d34e92aad3ca1480](https://www.sciencebase.gov/catalog/item/63acac09d34e92aad3ca1480). [Accessed: Oct. 7, 2026]

## 4. Location column descriptions

**Place:** Where the event actually occurred, but also where the event’s direct implications are felt most directly. When the researchers reviewed the articles, they were looking for mentions of specific places impacted by the events. Empty cell indicates a place was not coded for this event. NA indicates a place is not referenced in the event text.

```python
# Place column analysis:
# Number of each unique value
df['Place'].value_counts()
```

```python
# All unique values in column
df['Place'].unique()
```

**State:** State Name coded from Place field. Empty cell indicates a state was not coded for this event or that the article was not coded.

```python
# State column analysis:
df['State'].value_counts()
```

```python
# All unique values in column
df['State'].unique()
```

**Conflict Present:** Binary response to “Is a conflict present?” Conflict is defined here as an incompatibility of goals and various attempts of any present parties to control each other’s choices. When the researchers reviewed the articles, they were looking for evidence of this behavior in the event explained.

```python
# Conflict Present column analysis:
df['Conflict Present'].value_counts()
```

```python
# All unique values in column
df['Conflict Present'].unique()
```

## 9. Find unique state codes

```python
# Need to separate on spaces as well
# Select rows w/ conflict, state column
# Split that on ';'
# Explode to unlist that
# Remove ' ' from that
# Find unique string values of that
# (All other column values per row are duplicated)
state_series = (df.loc[df['Conflict Present'] == 'Y', 'State']
                .str.split(';')
                .explode()
                .str.strip(' ')
                .unique()
               )
```

```python
state_series
```

## 10. Visualize data exploration

```python
state_series_vc = (df.loc[df['Conflict Present'] == 'Y', 'State']
                .str.split(';')
                .explode()
                .str.strip(' ')
                .value_counts(dropna = False)
               )
state_series_vc
```

```python
state_series_vc.plot(kind = 'bar',
                     xlabel = 'State',
                     ylabel = 'Number of Conflicts',
                     title = 'Number of Conflicts per State')
```
